## Our First Statistics

The mean is the mathematical average set of numbers. 

```python
redshift.mean()
```

The standard deviation is a number that shows how spread out the values in a data set are from the average. The smaller the standard deviation, the tighter the points are spread across and are closer to the mean. This means the values are steady, reliable, and similar. A high standard deviation means the data points are spread far and show high variation. A zero standard deviation means there's no variation at all, and the numbers in the data set are identical.

```python
redshift.std()
```

```python
len(redshift)
```

Interpretation:  Our mean of about 1.72 represents the average redshift of quasars.
