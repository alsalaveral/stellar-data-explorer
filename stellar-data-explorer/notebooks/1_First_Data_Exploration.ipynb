## Indexing
The indexing operator and attribute selection are great because they work just like they do in the rest of the Python language. However, pandas has its own accessor operators, loc and iloc.

Looking at the table alone looks like a huge mess of information and can be overwhelming. An important skill in Python is being able to select the data points that are relevant to you. 

In Python, we can access the property of an object by accessing it as an attribute.

The indexing operator and attribute selection are great because they work just like they do in the rest of the Python language. However, pandas has its own accessor operators, loc and iloc.

Pandas indexing works in one of two ways. The first is an index-based selection where we select data based on its numerical position in the data. The way to do this is through iloc.

To select the first row of data in a DataFrame, we may:

```python
sdss17.iloc[0]
```

Both loc and iloc are row-first, column-second. This is the opposite of what we do in native Python, which is column-first, row-second. This means that it's easier to retrieve rows, but harder to retrieve columns. 

To get a column with iloc, we can do:

```python
sdss17.iloc[:,0]
```

The slice operator, or the : operator, when combined with other selectors, can be used to indicate a range of values. Because iloc means integer-location indexing, it selects rows and columns by their position (starting at 0). In this case, it's asked for all the rows but only for the first column. 

The general format of iloc is df.iloc[rows, columns]. If you provide only one argument, like we did the first example, pandas assumes you're talking about the rows.

The second way for attribute selection is the one followed by the loc operator, or label-based selection. In this, it's the data index value, not its position, which matters.

```python
str(sdss17.loc[0, 'obj_ID'])
```

When to Use Which?
- Use .loc when you have stable, meaningful index labels (like dates or IDs) and you want your code to remain human-readable.
- Use .iloc when you want to look up data purely by structure (e.g., getting the "first 5 rows" or "the last column") regardless of what the labels are

#### Conditional Selection

To do interesting things with the data, however, we often need to ask questions based on conditions.



For example, say we are interested in the objects that are classified as a quasar.

First, we can count how many values are in each category.

```python
sdss17["class"].value_counts()
```

In pandas, the most common way to filter rows is Boolean indexing.

```python
sdss17[sdss17["class"] == "QSO"]
```

```python
sdss17.loc[
    sdss17["class"] == "QSO"
]
```

```python
sdss17.loc[
    sdss17["class"] == "GALAXY",
    ['obj_ID', 'field_ID', 'class', 'redshift']
]
```

The term redshift measures how much the light from a distant celestial object is stretched to longer (redder) wavelengths as it travels through our expanding universe. The higher the redshift, the farther away the object is and the further back in time we are looking. A high red shift means that the object is further away and you are looking further back in time. So what if I wanted all the quasars with a red shift that is higher than 1?

```python
sdss17[(sdss17["class"] == "QSO")
    & (sdss17["redshift"] > 1)]
```

Or what if we wanted all galaxies that are between a certain range in red shift?

```python
sdss17[(sdss17["class"] == "GALAXY")
    & (sdss17["redshift"].between(1,1.5))]
```

Now suppose I want a table that have objects that are either quasars or galaxies, and have a high redshift, and don't all this unnecessary other information.

```python
sdss17[
    ((sdss17['class'] == "QSO") | (sdss17['class'] == "GALAXY")) &
    (sdss17['redshift'] > 1)
    ].loc[:, ['obj_ID', 'class', 'redshift']]
```

That's a lot of boolean expressions. We can simplify a little bit using a new method called .isin, and instead of hardcoding, we can use sequences.

```python
columns = ['obj_ID', 'class', 'redshift']
classes = ["QSO", "GALAXY"]
sdss17[
    (sdss17['class'].isin(classes)) &
    (sdss17['redshift'] > 1)
].loc[:, columns]
```

Filtering is important because real datasets are usually too large to analyze all at once. Researchers and data scientists filter data to focus on specific groups and answer targeted questions.

## Mapping
A map describes a function that takes one set of values and "maps" them to another set of values. In data science, we often  need to create new representations from existing data, or for transforming data from the format it is in now into the format that we want it to be in later on. Maps are what handle this work, making them extremely important for getting your work done!

### map()

map() takes every value in a column and replaces it with something else based on a rule. You can think of it as a translator. It's common pattern is df["column"].map(dictionary) or df["column"].map(function). Let's see it in action!

Suppose you have a table which has a categorical column, but the labels they use are hard to read. We can map each label to an easier one to read using a dictionary.

```python
class_names = {
    "GALAXY": "Galaxy",
    "STAR": "Star",
    "QSO": "Quasar"
}
sdss17["class"].map(class_names)
```

Our table, sdss17, has a column called 'MJD' or Modified Julian Date, and it's a way astronomers record when an observation happened using a single number. Astronomers use it because normal calendar dates (like "July 20, 2026") are inconvenient for calculations. A telescope might collect millions of observations, so researchers need a format that computers can easily compare and calculate. MJD is a count of days since a reference date. In this case, MJD 0 = November 17, 1858 (00:00 UTC). The larger the MJD, the more recent the observation.

```python
# What is the oldest observation?
oldest_date = sdss17["MJD"].min()
oldest_observation = sdss17["MJD"].idxmax()
print(f"Object #{oldest_observation} is the oldest observation with MJD {oldest_date}")
```

```python
# What is most recent observation?
recent_date = sdss17["MJD"].max()
recent_observation = sdss17["MJD"].idxmin()
print(f"Object #{recent_observation} is the most recent observation with MJD {recent_date}")
```

Suppose we want to categorize each object as an older or more recent observation.

```python
diff = recent_date - oldest_date
midpoint = oldest_date + diff / 2

# First define a function
def classify_observation(mjd):
    if mjd < midpoint:
        return "Early Survey"
    else:
        return "Late Survey"


sdss17["MJD"].map(classify_observation)
```

The function you pass to map() should expect a single value from the Series and return a transformed version of that value. map() returns a new Series where all the values have been transformed by your function.

### apply()

Another way to map is through apply(). apply() is the equivalent method if we want to transform a whole DataFrame by calling a custom method on each row.

The easiest way to think about it:
- map() → apply a function to each value in one column
- apply() → apply a function to a whole column or a whole row

The difference is when you want to use multiple columns together.

A photometric system defines the standardized filters, detectors, and mathematical scales used to measure the intensity of light. In astronomy, it classifies celestial objects based on brightness across specific ultraviolet, visible, and infrared wavelengths. In photometry systems, filters like the Red (R) filter let astronomers calculate an object's magnitude (a measure of brightness) in just red light. It tells you how much red light the object is emitting. Notice in our table we have multiple filters.

| Column | Filter | Measures |
|--------|--------|----------|
| u | Ultraviolet | Amount of UV light |
| g | Green | Amount of green visible light |
| r | Red | Amount of red visible light |
| i | Near-infrared | Amount of near-IR light |
| z | Infrared | Amount of IR light |

Now, suppose we want to classify objects based on what color they are. Astronomers often use differences between filters. Astronomers subtract the magnitude of the shorter-wavelength filter from the longer-wavelength filter. Because the Green (g) and Red (r) filters are both in the visible range and provide reliable measurements, we use the g-r color index to measure differences in the colors of astronomical objects. This allows us to compare how much light objects emit at shorter versus longer wavelengths. A smaller number indicates a hotter, bluer object, while a larger number indicates a cooler, redder one. 

```python
def classify_color(row):
    color_index = row["g"] - row["r"]

    if color_index < 0:
        return "Bluer Object"
    else:
        return "Redder Object"

sdss17["g_r_color"] = sdss17.apply(classify_color, axis=1)

sdss17["g_r_color"].value_counts()
```

Why axis=1?
- axis=0 → apply to each column
- axis=1 → apply to each row

### Reflection
I used apply() to create a new feature based on multiple columns. The g-r color index compares the brightness of an object in the green and red filters. A smaller color index generally indicates a bluer object, while a larger value indicates a redder object. I created a classification column to group objects based on their color index. However, real astronomy is more nuanced because the cutoff depends on what you are studying. A galaxy with g-r = 0.8 might be considered blue compared to an older galaxy. Astronomers often look at the distribution of colors instead, which we will do later on to try and more accurately classify.

## Grouping & Sorting
Grouping and sorting data are fundamental data techniques that transform raw, unstructured information into organized, readable, and actionable formats. They are great for information retrieval, spotting hidden trends, and simplifying complex data.

### groupby()
The groupby() method in pandas partitions a DataFrame into distinct groups to perform calculations. It executes the Split-Apply-Combine strategy: splitting data based on specified criteria, applying a function (like calculation or filtering), and combining the results into a new dataset. groupby() splits your data into groups based on one or more columns, then lets you perform calculations on each group.

The basic syntax is df.groupby("column") and is usually followed by an aggregation.

Suppose we want to know the average redshift for each object type?

```python
# Group by class, then work only with the redshift column.
sdss17.groupby("class")["redshift"].mean()
```

The first part sdss17.groupby("class") creates a GroupBy object where nothing is calculated yet. A GroupBy object is a special, intermediate container that holds your data segmented into distinct groups, but does not compute or display a new data table immediately. Pandas maps out which rows belong to which categories, but doesn't calculate anything until you explicitly tell it what operation to perform on those groups. In this case, our operation was mean()

```python
# You can work with multiple columns
sdss17.groupby("class")[["u", "g", "r"]].mean()
```

```python
# You can apply multiple aggregations too
sdss17.groupby("class")["redshift"].agg(["mean", "min", "max"])
```

MultiIndexes are something you'll start seeing once you use groupby() more. They look confusing at first, but they're just multiple labels used as an index instead of one.

Notice our past examples, our indices went from the generic 0, 1, 2, 3, 4... you see in the original dataframe to GALAXY, QSO, STAR. Now, suppose we group by two columns.

```python
sdss17.groupby(["class", "g_r_color"])["redshift"].mean()
```

Now pandas needs two labels to identify each row. Instead of treating class and g_r_color as normal columns, pandas makes them a MultiIndex. 

Multi-indices have several methods for dealing with their tiered structure which are absent for single-level indices. They also require two levels of labels to retrieve a value. Many beginners prefer normal columns.

```python
sdss17.groupby(["class", "g_r_color"])["redshift"].mean().reset_index()
```

Grouping returns data in index order, not in value order. When outputting the result of a groupby, the order of the rows is dependent on the values in the index, not in the data.

### concat()
For stacking tables, concat() is used to combine multiple pandas objects (DataFrames or Series) along a particular axis (vertically or horizontally). It basically stacks data on top of each other or side by side. It has five key parameters.

## Investigation
This is a good place to pause and use the pandas skills we learned to answer basic questions.

It has four parameters.
- objs → a list or mapping of DataFrames or Series to combine 
- axis → direction to concatenate (0 for vertical/rows, default, and 1 for horizontal/columns)
- join: how to handle mismatching indexes on the opposite axis ('outer' keeps all indexes, 'inner' keeps only overlapping indexes)
- ignore_index: if set to True, discards the original index markers and creates a continous label index from 0 to n-1
