## Introduction
An exoplanet is any planet that is located outside of our solar system. This ranges from gas giants, lava worlds, ice giants, or even terrestial planets similar to ours. As of 2026, astronomers have confirmed more than 6000 exoplanets out of the billions believed to be exist. However, the closest known exoplanet to is about 4 light-years away. 

There are various methods to discover exoplanets, and these are the main ones.
- The Wobble Method: A slight periodic shift in a star’s position or light frequency caused by the gravitational pull of an orbiting companion. A star’s radial velocity measures how fast it moves toward or away from us. If an exoplanet orbits the star, its gravity makes the star move back and forth, causing its radial velocity to periodically change. This repeating pattern can reveal the exoplanet.
- The Transit Method: A telescope, like the Kepler Space Telescope, watches for tiny, repeating drops in light when fixed on a small patch of stars. These tiny dips are caused by planets crossing in front of them. Once detected, the planet's orbital size and mass of the star can be calculated
- Direct Imaging: Telescopes take actual pictures of exoplanets by blocking the blinding glare of the host stars using specialized tools, helping us find young, massive, hot planets
- Gravitational Microlensing: Uses Einstein's theory of general relativity, in which gravity is able to warp and bend starlight. The gravity of a star in the foreground will bend and magnify the light of a distant background star when it passes in front

The dataset we will be working with is from the NASA Expolanet Archive, specifically the Planetary Systems (PS). According to NASA, in a single table view, PS combines data on exoplanets and their host starts, including discovery methods and stellar characteristics.

We can also stack queries.

```python
%%sql

SELECT
    pl_name,
    pl_rade,
    pl_bmasse
FROM exoplanets
WHERE pl_bmasse < 500 AND
    pl_rade IS NOT NULL
ORDER BY pl_rade ASC
LIMIT 5;
```

## Aggregation

Say we wanted to see the total mass of all the exoplanets that have a mass below 500. We'd use the sum() function.

```python
%%sql

SELECT SUM(pl_bmasse) AS total_mass
FROM exoplanets
WHERE pl_bmasse < 500 AND default_flag = 1;
```

We can also find the smallest, average, and largest value in a data set.

```python
%%sql

SELECT MIN(pl_bmasse) AS smallest__mass
FROM exoplanets;
```

```python
%%sql

SELECT MAX(pl_bmasse) AS largest_mass
FROM exoplanets
WHERE default_flag = 1;
```

```python
%%sql

SELECT AVG(pl_bmasse) AS average_mass
FROM exoplanets
WHERE default_flag = 1;
```

Say we want to find how many planets are in the database

```python
%%sql

SELECT COUNT(pl_name) AS planet_count
FROM exoplanets
WHERE pl_name IS NOT NULL;
```

But looking back at the data set, we can see there are repeats. That's why in our previous examples, we accounted for the default_flag, but let's say we didn't or we had a dataset without that.

```python
%%sql

SELECT COUNT(DISTINCT pl_name) AS planet_count
FROM exoplanets
WHERE pl_name IS NOT NULL;
```

We can stack multiple of these functions on top of each other.


```python
%%sql

SELECT  COUNT(DISTINCT pl_name) AS planet_count,
        AVG(pl_bmasse) AS average_mass,
        MIN(pl_bmasse) AS smallest_mass,
        MAX(pl_bmasse) AS largest_mass
FROM exoplanets
WHERE pl_name IS NOT NULL AND
      default_flag = 1;
```

We can group rows that have the same values into summary rows.

```python
%%sql
    
SELECT  discoverymethod,
        COUNT(pl_name) AS num_planets
FROM exoplanets
GROUP BY discoverymethod;
```

## Exercises

#### Exercise 1
Find the sum of the masses of the 5 smallest planets by mass.

```python
%%sql

SELECT SUM(pl_bmasse) AS total_mass
FROM (
    SELECT
        pl_name,
        pl_bmasse
    FROM exoplanets
    WHERE default_flag = 1
    ORDER BY pl_bmasse
    LIMIT 5
);
```

#### Exercise 2
Which three host stars have the most exoplanets?

```python
%%sql
SELECT *
FROM ( 
    SELECT  hostname,
            COUNT(pl_name) AS num_planets
    FROM exoplanets
    WHERE default_flag = 1
    GROUP BY hostname
)
ORDER BY num_planets DESC
LIMIT 3;
```

#### Exercise 3
How do the average masses of the planets in each discovery method differ from each other?

```python
%%sql

SELECT  discoverymethod,
        COUNT(pl_name) AS num_planets,
        AVG(pl_bmasse) AS average_mass
FROM exoplanets
WHERE default_flag = 1
GROUP BY discoverymethod
ORDER BY average_mass DESC;
```

#### Exercise 4
Which year had the most discoveries?

```python
%%sql

SELECT disc_year,
    COUNT(pl_name) AS num_planets
FROM exoplanets
WHERE default_flag = 1
GROUP BY disc_year
ORDER BY num_planets DESC;
```

#### Exercise 4
Which discovery methods have discovered at least 50 planets?

```python
%%sql

SELECT discoverymethod,
    COUNT(pl_name) AS num_planets
FROM exoplanets
WHERE (default_flag = 1) 
GROUP BY discoverymethod
HAVING COUNT(pl_name) >= 50
ORDER BY num_planets DESC;
```

#### Exercise 5
Classify planets with known orbital periods into ultra-short (under 1 day), short (1–10 days), medium (10–100 days), and long (over 100 days). How many planets fall into each category?

```python
%%sql
SELECT
    CASE
        WHEN pl_orbper < 1 THEN 'Ultra-Short'
        WHEN pl_orbper < 10 THEN 'Short'
        WHEN pl_orbper <= 100 THEN 'Medium'
        ELSE 'Long'
    END AS orbital_category,
    COUNT(*) AS num_planets
FROM exoplanets
WHERE pl_orbper IS NOT NULL 
    AND default_flag = 1
GROUP BY orbital_category
ORDER BY num_planets DESC;
```

#### Exercise 6
For each discovery method, calculate the total number of planets and the percentage with known mass and known radius (what proportion of planets discovered by a particular method have a recorded mass or radius measurement in the dataset).

```python
%%sql

SELECT
    discoverymethod,
    COUNT(*) AS total_planets,
    SUM(CASE
        WHEN pl_bmasse IS NOT NULL THEN 1
        ELSE 0
    END) AS known_mass,
    ROUND(
        100.0 * SUM(CASE
            WHEN pl_bmasse IS NOT NULL THEN 1
            ELSE 0
        END) / COUNT(*), 2) AS mass_percentage,
    SUM(CASE
        WHEN pl_rade IS NOT NULL THEN 1
        ELSE 0
    END) AS known_radius,
    ROUND(
        100.0 * SUM(CASE
            WHEN pl_rade IS NOT NULL THEN 1
            ELSE 0
        END) / COUNT(*), 2) AS radius_percentage
FROM exoplanets
WHERE default_flag = 1
GROUP BY discoverymethod
ORDER BY total_planets DESC;
```

#### Exercise 7
Compare the average radius and average mass of planets grouped into mass categories: under 2, 2–10, 10–100, and over 100 Earth masses.

```python
%%sql

SELECT
    CASE
        WHEN pl_bmasse < 2 THEN 'Small'
        WHEN pl_bmasse < 10 THEN 'Medium'
        WHEN pl_bmasse <= 100 THEN 'Large'
        ELSE 'Very large'
    END AS mass_category,

    COUNT(*) AS num_planets,
    ROUND(AVG(pl_bmasse), 2) AS avg_mass,
    ROUND(AVG(pl_rade), 2) AS avg_radius
FROM exoplanets
WHERE pl_bmasse IS NOT NULL
    AND default_flag = 1
GROUP BY mass_category;
```

#### Exercise 8
Which planets have orbital periods greater than the average orbital period of all planets with known periods?

```python
%%sql

SELECT
    AVG(pl_orbper) AS avg_orbper
FROM exoplanets
WHERE default_flag = 1;
```

```python
%%sql
SELECT 
    pl_name,
    hostname,
    pl_orbper
FROM exoplanets
WHERE pl_orbper> (
    SELECT AVG(pl_orbper)
    FROM exoplanets
    ) 
AND default_flag = 1;
```

#### Exercise 9
Which planets are larger than the average radius of the other planets in their own planetary system?

```python
%%sql
SELECT
    p.pl_name,
    p.hostname,
    p.pl_rade
FROM exoplanets AS p
WHERE p.pl_rade IS NOT NULL
AND p.default_flag = 1
AND p.pl_rade > (
    SELECT AVG(other.pl_rade)
    FROM exoplanets AS other
    WHERE other.hostname = p.hostname
      AND other.pl_name != p.pl_name
      AND other.pl_rade IS NOT NULL
)
ORDER BY p.hostname, p.pl_rade DESC;
```
